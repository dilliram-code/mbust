```python
# 2. Create variables of different data types(int, float, str, bool) and use the type() function to identify each one

# Integer variable
roll_no = 1
weight = 70.5
name = 'Mahendra'
is_student = True

print(type(roll_no))  
print(type(weight))
print(type(name))
print(type(is_student))	
```

```python
# 3. Write a program that takes two numbers as input from the user and prints the sum, difference, product and division

num1 = float(input("Enter the first number: "))
num2 = float(input("Enter the second number: "))

sum_result = num1 + num2
difference_result = num1 - num2
product_result = num1 * num2
division_result = (num1 / num2) if num2 != 0 else "Undefined!"

print(f"Sum: {sum_result}")
print(f"Difference: {difference_result}")
print(f"Product: {product_result}")
print(f"Division: {division_result: .2f}")
```

```python
# 4. Store your full name in a string variable and print its length, uppercase version, lowercase version, and replace spaces with underscores.

full_name = "Mahendra Chaudhary"
print(f"Length of full name: {len(full_name)}")
print(f"Uppercase version: {full_name.upper()}")
print(f"Lowercase version: {full_name.lower()}")
print(f"Full name with underscores: {full_name.replace(' ', '_')}")
```

```python
# 5. Create a list of 5 favorite movies, print each movie with its index number (starting from 1), then ask the user to add one more movie and display the final list.

favorite_movies = ["The Shawshank Redemption", "The Godfather", "The Dark Knight", "Pulp Fiction", "Forrest Gump"]

print("Favorite Movies:")
for index, movie in enumerate(favorite_movies, start=1):
  print(f"{index}. {movie}")
  
new_movie = input("Enter one more movie to add to the list: ")
favorite_movies.append(new_movie)

print("Updated Favorite Movies List:")
for index, movie in enumerate(favorite_movies, start=1):
  print(f"{index}. {movie}")
```

```python
# 6. Write a program that takes a sentence from the user and counts how many times the letter 'e' appears (case-insensitive).
sentence = input("Enter a sentence: ")
e_count = sentence.lower().count('e')
print(sentence)
print(f"The letter 'e' appears {e_count} times in the sentence.")
```
