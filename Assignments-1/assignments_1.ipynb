1. Install Python on your system and verify the installation by printing "Hello, World!" in both interactive mode and as a script file.

2. Create variables of different data types (int, float, str, bool) and use the type() function to identify each one.

3. Write a program that takes two numbers as input from the user and prints their sum, difference, product, and division.

4. Store your full name in a string variable and print its length, uppercase version, lowercase version, and replace spaces with underscores.

5. Create a list of 5 favorite movies, print each movie with its index number (starting from 1), then ask the user to add one more movie and display the final list.

6. Write a program that takes a sentence from the user and counts how many times the letter 'e' appears (case-insensitive).
