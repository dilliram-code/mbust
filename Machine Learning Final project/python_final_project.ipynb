```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns
import warnings 
warnings.filterwarnings('ignore')
```

#### Load dataset and get the overview of dataset

```python
original_dataset = pd.read_csv("city_day.csv")
df = original_dataset.copy()
```

```python
df.shape
```

##### Dataset has:

- 29, 531 samples
- 16 features

```python
df.head()
```

```python
# separate numerical and categorical columns 

numerical_cols = df.select_dtypes(exclude='object').columns
categorical_cols = df.select_dtypes(include='object').columns

print(f"Numerical cols: {numerical_cols}")
print("--"*50)
print(f"Categorical cols: {categorical_cols}")
```

```python
# check the null values by percentage

df.isnull().sum()/df.shape[0]*100
```

```python
# set the visual style

sns.set_theme(style='white')
```

```python
# plot AQI distribution: target distribution

plt.figure(figsize=(10, 6))
sns.histplot(df['AQI'].dropna(), bins=50, kde=True, color='blue')
plt.title('Distribution of Air Quality Index (AQI)')
plt.xlabel('AQI')
plt.ylabel('Frequency')
plt.tight_layout()
plt.show()
```

```python
# correlation heatmap 

corr_matrix = df[numerical_cols].corr()

# plot the heatmap

plt.figure(figsize=(12,8))
sns.heatmap(corr_matrix,annot=True, cmap='viridis', fmt=".2f", linewidths=0.5)
plt.title("Feature Correlation Heatmap")
plt.tight_layout()
plt.show()
```
